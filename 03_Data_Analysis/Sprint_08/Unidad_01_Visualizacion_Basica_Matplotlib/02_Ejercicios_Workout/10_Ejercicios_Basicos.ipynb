### Ejercicio 1

### Ejercicio 4

Reproduce la figura siguiente. Recuerda que las funciones seno hiperbólico y coseno hiperbólico se obtienen así:
* np.sinh, seno hiperbólico;
* np.cosh, coseno hiperbólico;


```python
fig = plt.figure()
ax1 = fig.add_axes([0.1, 0.7, 1, 0.5],
                   ylim=(-0, 10000))
ax2 = fig.add_axes([0.1, 0.01, 1, 0.5],
                   ylim=(0, 10000))

x = np.linspace(0, 10)
ax1.plot(np.sinh(x), color= "red")
ax2.plot(np.cosh(x));
ax1.set_title("Seno hiperbólico")
ax1.set_xlabel("x")
ax1.set_ylabel("sinh(x)");
ax2.set_title("Coseno hiperbólico")
ax2.set_xlabel("x")
ax2.set_ylabel("cosh(x)");
```

### Ejercicio 5: Ajustando el Espaciado de Subplots

Reproduce la siguiente figura con cuatro subplots, ajusta el espaciado entre los subplots para evitar la superosición (tal y como ves en la figura), emplea `subplots_adjust` y los parámetros hspace y wspace.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

fig, axs = plt.subplots(2, 2, figsize=(8,6))
fig.subplots_adjust(hspace=0.4, wspace=0.4)

axs[0,0].plot(x, np.sin(x))
axs[0,1].plot(x, np.cos(x))
axs[1,0].plot(x, np.tan(x))
axs[1,1].plot(x, np.exp(x))  

plt.show()
```


### Ejercicio 6: 

Reproduce la figura siguiente empleando sólo el estilo MATLAB (plt.\<funcion>\) y creando solo el contenido de los dos primeros gráficos (fila 1). Luego usa el estilo con métodos (objetos `axes`) para añadir el tercer gráfico.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

# Crear figura y primer eje (estilo MATLAB)
plt.figure(figsize=(8,6))

# Primer gráfico: coseno y raíz cuadrada
plt.subplot(2,1,1)
plt.plot(x, np.cos(x), label="coseno")
plt.plot(x, np.sqrt(x), label="raiz cuadrada")
plt.title("Raíz Cuadrada y Coseno")
plt.legend()

# Segundo gráfico: estilo orientado a objetos
ax = plt.subplot(2,1,2)
ax.plot(x, np.sin(x), label="seno", color='blue')
ax.set_title("Seno")
ax.legend()

plt.tight_layout()
plt.show()
```

Para terminar, reproduce la imagen de la figura

```python
x = np.linspace(0, 10, 100)  # Más puntos para suavidad
y = np.sin(x)

plt.figure(figsize=(8,4))
plt.plot(x, y,
         marker='o',
         linestyle='-',              # Línea continua
         color='blue',               # Color de línea            
         markerfacecolor='yellow',   # Relleno del marcador
         markeredgecolor='blue',     # Borde del marcador
         markersize=10,
         mew = 2,
         label='seno')

plt.title("Ejemplo de Marcadores")
plt.grid(True)
plt.legend()
plt.show()
```
