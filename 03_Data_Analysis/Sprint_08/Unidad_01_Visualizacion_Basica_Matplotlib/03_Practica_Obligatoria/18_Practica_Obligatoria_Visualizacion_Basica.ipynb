#### #1.2  

Recrea la siguiente figura, para ello construye en apartados separados cada gráfico y luego haz un apartado en el que se junten los cuatro.  

Las figuras son:
1. Las ciudades más pobladas, con los numeros de las poblaciones (tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center"). (Colores: azul, gris, aceituna, marrón, con trasparencia)

**1. Ciudades mas Pobladas**

```python
#1 : Ciudades mas Pobladas
fig, ax = plt.subplots(figsize=(8,6))

# Colores y datos
colors = ["blue", "grey", "salmon", "olive", "brown"]
top_cities = df.nlargest(5, "population_total").copy()
bars = ax.bar(top_cities["city"], top_cities["population_total"], color=colors, alpha=0.7)

# Texto en negro y arriba de cada barra
for bar, pop in zip(bars, top_cities["population_total"]):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height(),
            f"{int(pop):,}", ha="center", va="bottom", color="black", fontsize=11)

# Título y etiquetas
ax.set_title("Ciudades más pobladas de California")
ax.set_ylabel("Habitantes(miles)")
ax.set_xlabel("Ciudades")
plt.tight_layout()
plt.show()

```

```python

def grafico_ciudades_mas_pobladas(df,ax):

    # Colores y datos
    colors = ["blue", "grey", "salmon", "olive", "brown"]
    top_cities = df.nlargest(5, "population_total").copy()
    bars = ax.bar(top_cities["city"], top_cities["population_total"], color=colors, alpha=0.7)

    # Texto en negro y arriba de cada barra
    for bar, pop in zip(bars, top_cities["population_total"]):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height(),
                f"{int(pop):,}", ha="center", va="bottom", color="black", fontsize=11)

    # Título y etiquetas
    ax.set_title("Ciudades más pobladas de California")
    ax.set_ylabel("Habitantes (miles)")
    ax.set_xlabel("Ciudades")
```

**2. Histograma area total**

```python
fig, ax = plt.subplots(figsize=(8,6))

# Histograma con etiqueta para la leyenda
ax.hist(df["area_total_km2"], bins=50, color="steelblue", alpha=0.7, label="Hist Area")

# Ejes
ax.set_xlabel("Superficie (km²)",fontname="Bradley Hand ITC",fontsize= 18)
ax.set_ylabel("Número de Ciudades",fontname="Bradley Hand ITC",fontsize = 18)

# Mostrar la leyenda con fuente y tamaño
ax.legend(fontsize=10)

plt.tight_layout()
plt.show()
```

```python
def grafico_hist_area(df, ax):
    # Histograma con etiqueta para la leyenda
    ax.hist(df["area_total_km2"], bins=50, color="steelblue", alpha=0.7, label="Hist Area")

    # Ejes con fuente y tamaño
    ax.set_xlabel("Superficie (km²)", fontname="Bradley Hand ITC", fontsize=18)
    ax.set_ylabel("Número de Ciudades", fontname="Bradley Hand ITC", fontsize=18)

    # Mostrar la leyenda con fuente y tamaño
    ax.legend(fontsize=10)


```

**3. Poblacion por km2 de agua**

```python
fig, ax = plt.subplots(figsize=(8,6))

# Scatter: área de agua vs población total
ax.scatter(df["area_water_km2"], df["population_total"], 
           alpha=0.6, color="green")

# Títulos y etiquetas
ax.set_title("Población por kilómetro cuadrado de agua")
ax.set_xlabel("Área de agua en la ciudad (km²)")
ax.set_ylabel("Población (miles habitantes)")

# Punto con mayor población
ciudad_mayor_poblacion = df.loc[df["population_total"].idxmax()]

# Punto con mayor área de agua
ciudad_mayor_agua = df.loc[df["area_water_km2"].idxmax()]

# Flecha 1: señalar el outlier de población
ax.annotate("Outliers",
            xy=(ciudad_mayor_poblacion["area_water_km2"], ciudad_mayor_poblacion["population_total"]),
            xytext=(ciudad_mayor_poblacion["area_water_km2"] - 15, ciudad_mayor_poblacion["population_total"] - 500000),
            arrowprops=dict(arrowstyle="->", color="black", lw=2),
            )

# Flecha 2: conectar outlier de población con el de mayor área de agua
ax.annotate("",
            xy=(ciudad_mayor_agua["area_water_km2"], ciudad_mayor_agua["population_total"]),  # destino
            xytext=(ciudad_mayor_poblacion["area_water_km2"]-8, ciudad_mayor_poblacion["population_total"]- 500000),  # origen
            arrowprops=dict(arrowstyle="->", color="black", lw=2))

plt.tight_layout()
plt.show()

```

```python
def grafico_scatter_agua_poblacion(df, ax):
    # Scatter: área de agua vs población total
    ax.scatter(df["area_water_km2"], df["population_total"], 
               alpha=0.6, color="green")

    # Títulos y etiquetas
    ax.set_title("Población por kilómetro cuadrado de agua")
    ax.set_xlabel("Área de agua en la ciudad (km²)")
    ax.set_ylabel("Población (miles habitantes)")

    # Punto con mayor población
    ciudad_mayor_poblacion = df.loc[df["population_total"].idxmax()]

    # Punto con mayor área de agua
    ciudad_mayor_agua = df.loc[df["area_water_km2"].idxmax()]

    # Flecha 1: señalar el outlier de población
    ax.annotate("Outliers",
                xy=(ciudad_mayor_poblacion["area_water_km2"], ciudad_mayor_poblacion["population_total"]),
                xytext=(ciudad_mayor_poblacion["area_water_km2"] - 15, ciudad_mayor_poblacion["population_total"] - 500000),
                arrowprops=dict(arrowstyle="->", color="black", lw=2))

    # Flecha 2: conectar outlier de población con el de mayor área de agua
    ax.annotate("",
                xy=(ciudad_mayor_agua["area_water_km2"], ciudad_mayor_agua["population_total"]),
                xytext=(ciudad_mayor_poblacion["area_water_km2"] - 8, ciudad_mayor_poblacion["population_total"] - 500000),
                arrowprops=dict(arrowstyle="->", color="black", lw=2))
    
```

**4. Boxplot**

```python
fig, ax = plt.subplots(figsize=(8,6))

# Boxplot de la altitud
ax.boxplot(df["elevation_m"], vert=False, patch_artist=True,
           boxprops=dict(facecolor="red", color="black"),
           medianprops=dict(color="black", linewidth=1),
           whiskerprops=dict(color="black"),
           capprops=dict(color="black"),
           flierprops=dict(marker='o', markersize=6, linestyle='none'))

# Títulos y etiquetas
ax.set_title("Diagrama de caja de la Altitud de las ciudades de California")
ax.set_xlabel("Altitud (m)")

plt.tight_layout()
plt.show()

```

```python
def grafico_boxplot_altitud(df, ax):
    # Boxplot de la altitud
    ax.boxplot(df["elevation_m"], vert=False, patch_artist=True,
               boxprops=dict(facecolor="red", color="black"),
               medianprops=dict(color="black", linewidth=1),
               whiskerprops=dict(color="black"),
               capprops=dict(color="black"),
               flierprops=dict(marker='o', markersize=6, linestyle='none'))

    # Títulos y etiquetas
    ax.set_title("Diagrama de caja de la Altitud de las ciudades de California")
    ax.set_xlabel("Altitud (m)")

```

**COPIAR FIGURA**

```python
def plantilla_graficos(df):
    fig, axs = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle("Análisis de ciudades de California", fontsize=16)
    fig.subplots_adjust(hspace=0.4, wspace=0.3)

    grafico_ciudades_mas_pobladas(df, axs[0, 0])   # arriba izquierda
    grafico_hist_area(df, axs[0, 1])               # arriba derecha
    grafico_scatter_agua_poblacion(df, axs[1, 0])  # abajo izquierda
    grafico_boxplot_altitud(df, axs[1, 1])         # abajo derecha

    plt.tight_layout(rect=[0, 0, 1, 0.96])

    plt.savefig("analisis_ciudades_california.png", dpi=300)
    
    plt.show()

# Ejecutar
plantilla_graficos(df)

```
